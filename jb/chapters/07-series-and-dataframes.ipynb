# Series and DataFrames

A NumPy array keeps its values by position: the first value, the second value, and so on. Business data has
names: store S3, the column `sales`, the region `West`. **pandas** is the Python package for tables of data with
such names. In this chapter, we meet its two building blocks: the **Series** (one column) and the **DataFrame** (a
table).

**In this chapter, you will learn to:**

- create a Series with labels, and get its values and its labels;
- compute with a Series as with an array, and select a value by its label;
- create a DataFrame from a dictionary of lists;
- select one column (a Series) or several columns (a DataFrame);
- describe a DataFrame with `.shape`, `len()`, `.columns`, `.index` and `.dtypes`;
- add a new column to a DataFrame.

````python
import pandas as pd
````

By custom, pandas is imported as `pd`, just as NumPy is imported as `np`. pandas is built on NumPy: a pandas column
of numbers keeps its values in a NumPy array.

## Numbers need labels

A chain has six stores, S1 to S6. We know last month's sales of each store, in thousands of dollars. As a NumPy
array, the sales are six numbers: `[52.3, 48.1, 61.7, 39.5, 45.0, 53.4]`. Which number belongs to store S4? We
have to remember that the stores are in order, and count. With hundreds of stores, that is easy to get wrong.

pandas keeps a **label** with every value, so that we can ask for "the sales of S4" by name.

## A Series: values with labels

The simplest pandas object is a **Series**: a sequence of values, each with a label. We can make one from a list:

````python
monthly_sales = pd.Series([52.3, 48.1, 61.7, 39.5, 45.0, 53.4])
monthly_sales
````

> **In Thonny:** a last reminder. In a Thonny script, wrap the last line of a cell in `print()` to see its value,
> here `print(monthly_sales)`. From now on, the book takes this for granted.

pandas shows the Series in two columns. The right column holds the **values**. The left column, 0 to 5, is the
**index**: the labels of the values. We gave no labels, so pandas numbered the values from 0, like the positions
of an array. The last line, `dtype: float64`, is the type of the values, as for a NumPy array.

Now let's give the values better labels, and give the Series a name:

````python
monthly_sales = pd.Series(
    [52.3, 48.1, 61.7, 39.5, 45.0, 53.4],
    index=["S1", "S2", "S3", "S4", "S5", "S6"],
    name="sales",
)
monthly_sales
````

Now the index holds the store names, and the last line also shows the name of the Series, `sales`. `index=` takes
a list of labels, one for each value, in the same order. `name=` is optional. It becomes the column name when a
Series is part of a table.

**The rule:** a Series is a column of values with an **index** of labels. All its values have one dtype, as in a
NumPy array.

## The values and the index

A Series has two parts, and we can get each one. `.to_numpy()` gives the values as a NumPy array:

````python
monthly_sales.to_numpy()
````

This is the kind of array we used in the chapter *From lists to NumPy arrays*: the same six values, without the
labels. You may also see `.values` in other code. For a column of numbers, it gives the same array. We use
`.to_numpy()`, whose name says what it does.

`.index` gives the labels:

````python
monthly_sales.index
````

The labels form a pandas object of their own, an `Index`. Its dtype is `'object'`, which is how pandas 2 marks
text. In pandas 3, the dtype is `'str'`. We come back to dtypes below.

## Arithmetic works as with arrays

A Series computes like a NumPy array: an operation with a number is done on every value. Say that each store must
grow its sales by 10%. The target of each store is:

````python
monthly_sales * 1.10
````

Each value was multiplied by 1.10, and **the labels stayed**: we can still see which target belongs to which store.
The summaries of the chapter *Comparisons and Boolean masks*, such as `.sum()`, `.mean()` and `.max()`, work too.

> **Try it:** run `monthly_sales > 50`. You get a Series of `True` and `False`: a mask, with labels. Which stores
> sold more than 50 thousand dollars? (The chapter *Filtering rows* uses such masks to choose rows.)

## Selecting a value by its label

To get one value, we put its label in square brackets, as with a key in a dictionary:

````python
monthly_sales["S3"]
````

Store S3 sold 61.7 thousand dollars. A Series is a little like a dictionary whose keys are the labels, but it can
also compute like an array.

> **Note:** Colab shows `np.float64(61.7)`, and `print()` shows `61.7`: the NumPy form of a number, as in the
> NumPy chapters. pandas keeps its numbers in NumPy arrays, so one number taken out of pandas is a NumPy number.
> A whole number shows as `np.int64(...)` in the same way.

> **Common mistake:** writing `monthly_sales[0]` to get the first value. The square brackets of a Series look for a
> **label**, and `0` is not one of the labels here. Colab's pandas 2 still gives the first value, with a
> `FutureWarning`; pandas 3 gives `KeyError: 0`. To select by position, use `iloc`, which the chapter
> *Selecting with loc and iloc* explains.

## A DataFrame: a table of columns

Most business data is a table: one row for each store (or customer, or sale), and one column for each fact. pandas
calls a table a **DataFrame**. The easiest way to make one is from a dictionary: each key is a column name, and
each value is the list of the values in that column.

````python
stores = pd.DataFrame({
    "store": ["S1", "S2", "S3", "S4", "S5", "S6"],
    "city": ["Boston", "Atlanta", "Chicago", "Denver", "Miami", "Seattle"],
    "region": ["East", "South", "North", "West", "South", "West"],
    "employees": [12, 8, 15, 6, 9, 10],
    "sales": [52.3, 48.1, 61.7, 39.5, 45.0, 53.4],
})
stores
````

Each row is one store, and each column is one fact about it: its name, its city, its region, its number of
employees, and its monthly sales in thousands of dollars. On the left is the index, 0 to 5. We gave no index, so
pandas numbered the rows.

**The rule:** a DataFrame is a table of columns that **share one index**. Each column is a Series. Different
columns can hold different types (text in one, numbers in another), but all the values in one column have one
dtype. All the lists must have the same length: one value for each row.

## One column is a Series

To select a column, we put its name in square brackets:

````python
stores["sales"]
````

The result is a **Series**: the values of the column, with the index of the DataFrame on the left, and the column
name as the Series' name. Everything we did with `monthly_sales` works here too: `stores["sales"].mean()` is the
average sales of a store, and `stores["employees"].sum()` the number of employees of the whole chain.

> **Common mistake:** `stores["Sales"]`, with a capital S, gives `KeyError: 'Sales'`. Column names are
> case-sensitive: write them exactly as they are in the table.

## Several columns: double brackets

To select several columns, we put a **list** of column names inside the square brackets. That makes two pairs of
brackets: the outer pair selects, and the inner pair makes the list.

````python
stores[["city", "sales"]]
````

The result is a **DataFrame** with only these two columns, in the order of the list. Even a list with one name
gives a DataFrame: `stores[["sales"]]` is a table with one column, while `stores["sales"]` is a Series. (The
chapter *What does a model learn?* wrote `customers[["visits_per_month"]]` for this reason: scikit-learn wants the
predictors as a table.)

What happens if we forget the inner brackets?

````python
stores["city", "sales"]
````

The last line of the message is `KeyError: ('city', 'sales')`. Without the inner brackets, Python passes the two
names together as one tuple, and pandas looks for **one** column whose name is that tuple. There is no such column.
The fix is `stores[["city", "sales"]]`.

## The size and the parts of a table

`.shape` gives the size of a DataFrame as (rows, columns), as for a 2-D array in the chapter
*Two-dimensional arrays*. `len()` gives the number of rows:

````python
print(stores.shape)
print(len(stores))
````

The table has 6 rows and 5 columns. For a DataFrame, `len` counts the **rows**, not the columns.

`.columns` gives the column names, and `.index` the row labels:

````python
print(stores.columns)
print(stores.index)
````

The column names are an `Index` of text, like the labels of `monthly_sales` (`dtype='object'`, or `dtype='str'` in
pandas 3). We can loop over them, or make a list with `list(stores.columns)`.

The row labels are a `RangeIndex`: pandas' short way to write the numbers 0, 1, ..., 5. It starts at 0 and stops
**before** 6, like `np.arange(6)`. When we read a data file in the next chapter, its rows get such an index too.

## The type of each column

`.dtypes` (with an `s`) gives the dtype of every column:

````python
stores.dtypes
````

The number columns have NumPy's dtypes: `int64` for whole numbers (`employees`) and `float64` for numbers with
decimals (`sales`). The text columns `store`, `city` and `region` have the dtype `object` (or `str` in pandas 3):
that is how pandas marks text. Ignore the last line, `dtype: object`: it is the dtype of this list of dtypes
itself, and it is the same in both versions.

Why look at dtypes? Arithmetic and summaries work only on number columns. And machine-learning models accept only
numbers: in the chapter *A first machine-learning example*, `fit` failed because of the text column `region`.
`.dtypes` is the quickest way to find such columns.

## Adding a column

To add a column, we assign a list to a new column name. The list needs one value for each row. Here is the year in
which each store opened:

````python
stores["opened"] = [2012, 2018, 2009, 2021, 2016, 2019]
stores
````

The new column appears at the right. (If a column with that name already exists, the assignment replaces its
values.)

More often, a new column is computed from other columns. The arithmetic works row by row, as for two arrays of the
same length. Here are the sales of each store divided by its number of employees:

````python
stores["sales_per_employee"] = stores["sales"] / stores["employees"]
stores
````

For each row, pandas divided that row's sales by that row's employees. Store S3 in Chicago has the highest sales
(61.7) and the most employees, but the lowest sales per employee: about 4.11 thousand dollars a month. Store S4 in
Denver has the lowest sales and the fewest employees, but the highest sales per employee: about 6.58. The chapter
*New columns and sorting* shows more ways to make columns.

> **Common mistake:** a list with the wrong number of values. `stores["opened"] = [2012, 2018]` gives
> `ValueError: Length of values (2) does not match length of index (6)`. The list must have exactly one value for
> each row.

## Summary

- **pandas** (`import pandas as pd`) works with tables of data. It is built on NumPy.
- A **Series** is a column of values with an **index** of labels: `pd.Series([...], index=[...], name="...")`.
  `.to_numpy()` gives its values as a NumPy array, and `.index` its labels. Arithmetic and summaries work as for
  arrays, and `s["S3"]` selects a value by its label.
- A **DataFrame** is a table of columns that share one index: `pd.DataFrame({"col": [...], ...})`.
- `df["col"]` gives one column, as a Series. `df[["a", "b"]]` (double brackets) gives a DataFrame with those
  columns.
- `df.shape` is (rows, columns), `len(df)` is the number of rows, `df.columns` gives the column names, and
  `df.index` the row labels.
- `df.dtypes` gives the type of each column: `int64` and `float64` for numbers, `object` (or `str` in pandas 3) for
  text.
- `df["new"] = [...]` or `df["new"] = df["a"] / df["b"]` adds a column.

### Question 7 (Challenge)

A colleague types the number of employees of four new stores, but one value goes in as text:
`new_employees = pd.Series([12, 8, "15", 6])`. What is the dtype of this Series? What happens if you call
`new_employees.sum()`, and why? How do you fix the data?

<details class="yr-answer">
<summary>Answer</summary>

The dtype is `object`, in pandas 2 and in pandas 3: the Series mixes numbers and text, so it cannot be `int64`
(and pandas 3 uses `str` only when every value is text). `.dtype` gives the dtype of one Series, as for an array.
`new_employees.sum()` fails with `TypeError: unsupported operand type(s) for +: 'int' and 'str'`: Python cannot add
a number and a text. The fix is to write `15` without quotes. Then the dtype is `int64`, and the sum is 41.

```python
new_employees = pd.Series([12, 8, "15", 6])
print(new_employees.dtype)
fixed = pd.Series([12, 8, 15, 6])
print(fixed.dtype)
print(fixed.sum())
```

</details>
