## The business question

A company sells one product in four regions. Every week, in each region, it decides how much money to spend on TV
advertising and on online advertising, sets the price, and sometimes runs a promotion in the stores. Then it counts
the units sold.

The marketing director asks two questions:

1. **Can we predict the units sold in a week** from the advertising, the price, the promotion and the region? A
   good prediction helps to plan stock and budgets.
2. **Which kind of advertising seems to matter**, TV or online?

## The data

The file `marketing.csv` has one row for each week and region, 520 rows in all (130 weeks in four regions). The
appendix *The datasets* describes it:

- `region`: `North`, `South`, `East` or `West` (text);
- `tv_spend` and `online_spend`: the money spent on TV and online advertising that week, in **thousands** of
  dollars;
- `price`: the price of the product that week, in dollars;
- `promotion`: `Yes` or `No` (text): whether a store promotion ran that week;
- `units_sold`: the number of units sold that week. This is the **target**.

The model solution below drops the rows. Filling instead keeps all 520 weeks, and linear regression is still the

filled = marketing.copy()

filled = pd.get_dummies(filled, columns=["region", "promotion"], dtype=int)

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error

DATA_URL = "https://raw.githubusercontent.com/y-rosenthal/yr-pre-ml/main/data/"

# Step 1: read and inspect
marketing = pd.read_csv(DATA_URL + "marketing.csv")
print(marketing.shape)
marketing.info()
print(marketing.describe().round(1))

# Step 2: explore
print(marketing["region"].value_counts())
print(marketing["promotion"].value_counts())
by_region = marketing.groupby("region")["units_sold"].mean().round(1)
print(by_region.sort_values())
print(marketing.groupby("promotion")["units_sold"].mean().round(1))
for column, cut in [("tv_spend", 25), ("online_spend", 10)]:
    high = marketing.loc[marketing[column] > cut, "units_sold"].mean()
    low = marketing.loc[marketing[column] <= cut, "units_sold"].mean()
    print(f"{column} above {cut}: {high:.1f} units; {cut} or less: {low:.1f} units")

by_region.plot(kind="bar")
plt.title("Average weekly units sold by region")
plt.xlabel("region")
plt.ylabel("average units sold")
plt.show()

# Step 3: clean. The company sets its advertising, price and promotion itself,
# so a future week has all its values. The missing values are only gaps in the
# records of past weeks, and dropping them keeps most of the weeks.
print(marketing.isna().sum())
clean = marketing.dropna()
print("weeks left:", len(clean))

# Step 4: text to numbers
clean = pd.get_dummies(clean, columns=["region", "promotion"], dtype=int)

# Step 5: X and y
X = clean.drop(columns="units_sold")
y = clean["units_sold"]
print(X.shape, y.shape)
print("average units sold per week:", round(y.mean(), 1))

# Step 6: split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
print(X_train.shape, X_test.shape)

# Steps 7 and 8: three models and their MAE
linear = LinearRegression()
linear.fit(X_train, y_train)
tree = DecisionTreeRegressor(max_depth=5, random_state=42)
tree.fit(X_train, y_train)
forest = RandomForestRegressor(n_estimators=200, random_state=42)
forest.fit(X_train, y_train)
for name, model in [("linear regression", linear), ("decision tree", tree), ("random forest", forest)]:
    error = mean_absolute_error(y_test, model.predict(X_test))
    print(f"{name:18} MAE: {error:.1f} units")

# Step 9: which advertising matters? Train the best model without each one.
for column in ["tv_spend", "online_spend"]:
    model = LinearRegression()
    model.fit(X_train.drop(columns=column), y_train)
    error = mean_absolute_error(y_test, model.predict(X_test.drop(columns=column)))
    print(f"without {column:13} MAE: {error:.1f} units")
print(pd.Series(linear.coef_, index=X.columns).round(2))
```

<details class="yr-answer">
<summary>Answer</summary>

Linear regression predicts the weekly units sold best: on the test weeks, it is off by 28.6 units on average, about
7% of an average week (426.7 units), which is precise enough to plan stock. The random forest (39.1 units) and the
decision tree (62.3 units) are clearly worse. Both kinds of advertising matter: without TV spending, the error of
linear regression grows to 80.2 units, and without online spending to 70.6 units. For each thousand dollars,
online advertising seems to bring more: 14.32 extra units a week, against 5.84 for TV. Promotions help too (466.4
units on average in promotion weeks, against 409.7), and each extra dollar of price costs about 17 units (the
model's number is -17.13). These are patterns in past weeks, not proof, so we suggest moving a part of the TV
budget to online advertising for a few weeks and measuring the result.
