## How to read an error message

When Python cannot run a line, it stops and shows an **error message**. A long message is called a *traceback*:
it lists the lines that Python was running when the problem happened, from your line down into the code of pandas
or scikit-learn. Most of it is meant for experts. Two parts matter:

- **The last line.** It has the form `ErrorType: message`. The type names the kind of problem (`NameError`,
  `KeyError`, `ValueError`, ...), and the message gives the details.
- **Your line.** In Colab, an arrow (`---->`) points at the line of your cell that failed. In Thonny, look for the
  line of the message that names your file.

Each section below shows an error on purpose, explains its last line, and then shows the fix. The website shows
the messages of Colab's versions of Python and the packages. Other versions may word a message a little
differently, but the kind of error is the same.

> **Note:** an error does no harm. Fix the cell and run it again. In Colab, **Runtime > Run all** stops at the first
> cell with an error, so fix that cell first.

## NameError: a cell that did not run

Suppose that we open this notebook and run a cell in the middle, without running the cells above it:

````python
sales.head()
````

The last line is `NameError: name 'sales' is not defined`. A **name** is the name of a variable or a function.
Python knows only the names that the code it already ran has created. No cell has created `sales` yet: the cell
that reads the file did not run.

This happens a lot in notebooks. When you open a notebook again, or when Colab restarts after some time without
use, every variable is gone, even though the cells may still show their old outputs.

**The fix:** run the cell that creates the name, and the cells above it. In Colab, **Runtime > Run before** runs
every cell above the current one. Here, we read the file:

````python
sales = pd.read_csv(DATA_URL + "sales.csv")
sales.head()
````

Now `sales` exists, and `head()` shows the first five sales.

The last line is `NameError: name 'sale' is not defined`. Python does not guess what we meant: `sale` and `sales`
are two different names, and so are `Sales` and `sales` (names are case-sensitive). When Python runs a script, for
example in Thonny, it may add a hint to the message: `Did you mean: 'sales'?`.

**The fix:** correct the spelling.

````python
sales["units"].sum()
````

The stores sold 8,949 units in all.

## SyntaxError and IndentationError: code that is not Python

Some errors appear before any line runs. Python first reads the whole cell. If the code breaks the rules of the
language, Python stops at once with a `SyntaxError`, and **no line of the cell runs**. (We show these errors as
text, not as cells that run.)

A `for` line without the colon at its end:

```python
for price in [4.99, 12.50, 3.25]
    print(price)
```

```
    for price in [4.99, 12.50, 3.25]
                                    ^
SyntaxError: expected ':'
```

A missing closing parenthesis:

```python
print(sales.head()
```

```
SyntaxError: '(' was never closed
```

A line that should be indented, but is not:

```python
for price in [4.99, 12.50, 3.25]:
print(price)
```

```
IndentationError: expected an indented block after 'for' statement on line 1
```

The caret `^` (or `^^^^^`) marks the place where Python noticed the problem. The mistake is there or just before
it, sometimes at the end of the line above.

**The fix:** add the missing `:`, `)` or quote, or indent the lines of the block (four spaces).

## TypeError: adding text and a number

We want to show a total with a label:

````python
revenue = 1234.5
"Revenue: $" + revenue
````

The last line is `TypeError: can only concatenate str (not "float") to str`. A **type** is the kind of a value:
`str` (text), `int` (a whole number), `float` (a number with a decimal point). `+` joins two strings (it
*concatenates* them), or adds two numbers, but it cannot do both at once. Python does not guess whether we want
text or a number.

**The fix:** turn the number into text with `str()`, or, better, use an f-string:

````python
print("Revenue: $" + str(revenue))
print(f"Revenue: ${revenue:,.2f}")
````

Both lines work. In the f-string, `:,.2f` shows the number with a comma between the thousands and two decimals:
`$1,234.50`.

> **Note:** the other order, `100 + "5"`, gives a different message for the same problem:
> `TypeError: unsupported operand type(s) for +: 'int' and 'str'`.

## KeyError: a misspelled column name

````python
sales["Store"]
````

This traceback is long: pandas shows two errors, the second caused by the first ("The above exception was the
direct cause of the following exception"). Read the very last line: `KeyError: 'Store'`. A **key** is a label
that we look up, as in a Python dictionary. pandas looked for a column named `Store` and found none: the column is
`store`, in lowercase. Column names are case-sensitive, and a space counts too (`"store "` is another name).

**The fix:** look at the column names, and copy the name exactly:

````python
print(sales.columns)
sales["store"].nunique()
````

The list of columns shows `'store'` in lowercase (the `dtype` at its end is `object`, or `str` in pandas 3). With

## ValueError: arrays of different lengths

We have four sales: the units of each sale, and the price of one unit. But one price is missing:

````python
units = np.array([3, 1, 4, 2])
prices = np.array([2.50, 10.00, 4.00])
units * prices
````

The last line is `ValueError: operands could not be broadcast together with shapes (4,) (3,)`. The **operands**
are the two arrays that `*` works on. **Broadcasting** is NumPy's word for matching up the values of two arrays so
that it can compute element by element. The shape `(4,)` means 4 values in one row, and `(3,)` means 3 values.
NumPy cannot pair 4 units with 3 prices.

A `ValueError` means that a value has the right type but a wrong value: here, two arrays, but of lengths that do
not fit together.

**The fix:** find the missing value, so that both arrays have one value for each sale. `.shape` (or `len()`)
shows the lengths:

````python
prices = np.array([2.50, 10.00, 4.00, 7.50])
print(units.shape, prices.shape)
units * prices
````

Both arrays have 4 values now, and we get the revenue of each sale.

## ValueError: `and` instead of `&`

We want the big sales (more than 10 units) in the North region. There are two conditions, so we write `and`:

````python
sales[sales["units"] > 10 and sales["region"] == "North"]
````

The last line is `ValueError: The truth value of a Series is ambiguous. Use a.empty, a.bool(), a.item(), a.any()
or a.all().` (pandas 3 leaves `a.bool()` out of the list.) Python's `and` needs a single `True` or `False` on each
side. But `sales["units"] > 10` is a Series of 2,000 values, one for each sale, and Python cannot turn 2,000 values
into one `True` or `False`: that is what "ambiguous" means. The suggestions in the message (`a.any()`, `a.all()`,
...) are for other situations. They are not what we want here.

**The fix:** combine conditions with `&` (and), `|` (or) and `~` (not), and put each condition in parentheses
(the chapter *Filtering rows*):

````python
big_north = sales[(sales["units"] > 10) & (sales["region"] == "North")]
len(big_north)
````

27 sales match both conditions. `or` and `not` give the same error: use `|` and `~` instead.

> **Common mistake:** forgetting the parentheses, as in `sales["units"] > 10 & sales["region"] == "North"`. Python
> computes `&` before `>` and `==`, so this means something else, and it stops with a `TypeError` whose message is
> hard to understand. Always put each condition in parentheses.

## A warning: chained assignment

Not every problem stops the code. A **warning** is a message that Python shows while the code goes on. Here we
copy `sales`, and try to set the discount of store S1 to 0.5:

```python
sales2 = sales.copy()
sales2[sales2["store"] == "S1"]["discount"] = 0.5
```

(We show this code as text, because its warning differs between pandas 2 and pandas 3.) In Colab (pandas 2), the
warning says:

```
SettingWithCopyWarning:
A value is trying to be set on a copy of a slice from a DataFrame.
Try using .loc[row_indexer,col_indexer] = value instead
```

In pandas 3, it starts like this (despite its name, it is a warning too):

```
ChainedAssignmentError: A value is being set on a copy of a DataFrame or Series through chained assignment.
```

In both versions, **`sales2` does not change**. The line has two steps in a chain: `sales2[sales2["store"] ==
"S1"]` makes a new, temporary table with the rows of S1, and `["discount"] = 0.5` changes that temporary table,
which is then thrown away.

**The fix:** do it in one step, with `loc` (the chapter *Filtering rows*):

````python
sales2 = sales.copy()
sales2.loc[sales2["store"] == "S1", "discount"] = 0.5
sales2.loc[sales2["store"] == "S1", "discount"].unique()
````

The only discount left in the rows of S1 is 0.5: the change worked, with no warning.

## ValueError from fit: text in X

In machine learning, many errors come from `fit`. Here we try to predict `annual_spend` from the visits and the
region:

````python
X = customers[["visits_per_month", "region"]]
y = customers["annual_spend"]
model = LinearRegression()
model.fit(X, y)
````

The last line is `ValueError: could not convert string to float: 'South'`. A model computes with numbers
(`float`), so `fit` tries to turn every value of `X` into a number. `'South'`, the region of the first customer, is
text, and it cannot become a number.

**The fix:** turn each text column into number columns with `pd.get_dummies` (the chapter
*Preparing data for machine learning*):

````python
X = customers[["visits_per_month", "region"]]
X = pd.get_dummies(X, columns=["region"], dtype=int)
model = LinearRegression()
model.fit(X, y)
````

No error this time: instead of an error message, Colab shows a short description of the model, so `fit` worked.
`X` now has the columns `visits_per_month`, `region_East`, `region_North`, `region_South` and `region_West`, all
numbers.

## ValueError from fit: missing values

Now we try the visits and the age:

````python
X = customers[["visits_per_month", "age"]]
y = customers["annual_spend"]
model = LinearRegression()
model.fit(X, y)
````

This message has two lines at the end. The first one, `ValueError: Input X contains NaN.`, says the problem: `X`
has missing values (`NaN`). The `age` column has 14 of them. The last line is long advice from scikit-learn, which
starts with `LinearRegression does not accept missing values encoded as NaN natively.` It suggests other models
and tools; in this book, we prepare the data instead.

**The fix:** drop the rows with a missing value, or fill them in (the chapter
*Preparing data for machine learning*).
Here we drop them. The rows of `X` and `y` must stay together, so we first drop the rows from the table, and then
take `X` and `y` from the result:

````python
clean = customers.dropna(subset=["age"])
X = clean[["visits_per_month", "age"]]
y = clean["annual_spend"]
print(X.shape, y.shape)
model = LinearRegression()
model.fit(X, y)
````

`X` and `y` both have 486 rows, and `fit` works.

> **Common mistake:** dropping the missing values from `X` only, as in
> `X = customers[["visits_per_month", "age"]].dropna()` with `y = customers["annual_spend"]`. Then `X` has 486 rows
> and `y` has 500, and `fit` stops with `ValueError: Found input variables with inconsistent numbers of samples:
> [486, 500]`. Take `X` and `y` from the same clean table.

## When you are stuck

1. **Read the last line**: the type of the error and its message.
2. **Find your line**: the arrow in Colab, or the line with your file name in Thonny.
3. **Check the names**: the spelling and the capitals of variables, columns (`df.columns`) and files.
4. **Check what you really have**: `type(x)`, `x.shape`, `df.dtypes` and `df.isna().sum()` show it.
5. **Run the notebook from the top** with **Runtime > Run all**. Cells that ran out of order can leave variables
   that the code above them does not explain.
6. **Search the web for the last line** (without your own variable names), or ask someone. Show them the whole
   message, and the code that gave it.

## Summary

- Read the **last line** of an error message first: `ErrorType: message`.
- `NameError`: no code created that name. Run the cells above, or correct the spelling.
- `ModuleNotFoundError`: a misspelled package name, or a package that is not installed.
- `SyntaxError` and `IndentationError`: the code is not valid Python, and no line of the cell runs.
- `TypeError` with `+`: text and a number. Use an f-string, `f"Revenue: ${revenue}"`.
- `HTTPError: HTTP Error 404: Not Found`: a misspelled data file name. `FileNotFoundError`: a file on your computer
  that is not in the script's folder.
- `KeyError`: a misspelled column name. Check `df.columns`.
- `ValueError: operands could not be broadcast together`: arrays of different lengths.
- `ValueError: The truth value of a Series is ambiguous`: `and`, `or`, `not` instead of `&`, `|`, `~`.
- `SettingWithCopyWarning` (pandas 2) or `ChainedAssignmentError` (pandas 3): chained assignment, which changes
  nothing. Use `df.loc[mask, "col"] = value`.
- `fit` errors: `could not convert string to float` (use `pd.get_dummies`) and `Input X contains NaN` (use
  `dropna` or `fillna`, on the table before you take `X` and `y`).
