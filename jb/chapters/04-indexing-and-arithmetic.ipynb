## The shop's data

A small shop sells six products. We keep their prices in one array, and the units sold last week in another. The
two arrays are in the same order, so the same position means the same product in both:

| Position | Product | Price ($) | Units sold |
|---|---|---|---|
| 0 | coffee beans | 12.00 | 10 |
| 1 | tea | 4.50 | 25 |
| 2 | chocolate | 7.25 | 8 |
| 3 | coffee grinder | 30.00 | 2 |
| 4 | cookies | 2.75 | 40 |
| 5 | mug | 15.50 | 6 |

````python
prices = np.array([12.00, 4.50, 7.25, 30.00, 2.75, 15.50])
units = np.array([10, 25, 8, 2, 40, 6])
prices
````

> **In Thonny:** wrap the last line in `print()`: `print(prices)`. Thonny then shows the values without the word
> `array` and without commas: `[12.    4.5   7.25 30.    2.75 15.5 ]`.

Colab shows the six prices. Notice that the positions in the table start at 0, not at 1: arrays count positions
the same way as Python lists.

## One element: indexing

To get one element, write its position in square brackets after the array's name. The position is called the
**index**. The coffee beans are at position 0:

````python
prices[0]
````

Colab shows `np.float64(12.0)`. This is the number 12.0, stored as a NumPy `float64` (the dtype of the array, from
the previous chapter). In Thonny, `print(prices[0])` shows just `12.0`. We can use it like any other number. For
example, here is the revenue of the coffee grinder, at position 3:

````python
prices[3] * units[3]
````

30.00 × 2 = 60.0 dollars. A **negative index** counts from the end: `-1` is the last element, `-2` the one before
it, and so on. The last product is the mug:

````python
prices[-1]
````

With `-1`, we do not need to know how long the array is.

> **Common mistake:** writing `prices[6]` for the sixth product. The positions are 0 to 5, so the sixth product is
> `prices[5]` (or `prices[-1]`). `prices[6]` gives the error `IndexError: index 6 is out of bounds for axis 0 with
> size 6`: there is no position 6 in an array of size 6. (Axis 0 is the one direction of a 1-D array; the
> chapter *Two-dimensional arrays* explains axes.)

## Several elements: slicing

To get several neighboring elements, write a **slice**: a start and a stop, with a colon between them. Here are the
prices at positions 1, 2 and 3:

````python
prices[1:4]
````

A slice gives a new, shorter array: the prices of the tea, the chocolate and the grinder. As with `np.arange` and
with list slices, **the stop is not included**: `1:4` means the positions 1, 2 and 3.

Leave out the start to begin at the first element. Here are the first three prices:

````python
prices[:3]
````

Leave out the stop to go to the end. `prices[3:]` gives the prices from position 3 to the end, the last three.

A slice can also have a third number, the **step**, as in `start:stop:step`. With a step of 2, we get every second
element. Here we leave out the start and the stop, so the slice covers the whole array:

````python
prices[::2]
````

The elements at positions 0, 2 and 4: the coffee beans, the chocolate and the cookies.

| Slice | Positions | Meaning |
|---|---|---|
| `a[1:4]` | 1, 2, 3 | from 1 up to (not including) 4 |
| `a[:3]` | 0, 1, 2 | the first three |
| `a[3:]` | 3, 4, 5 | from 3 to the end |
| `a[::2]` | 0, 2, 4 | every second element |

## Changing an element

We can also change an element: put the index on the left of `=`. Another shop, a gift shop, keeps its prices in
whole dollars. The price of its third product drops from 15 to 12 dollars:

````python
gift_prices = np.array([25, 40, 15, 60])
gift_prices[2] = 12
gift_prices
````

The element at position 2 is now 12. The other elements did not change.

Now the gift shop changes its first price from 25 to 24.99:

````python
gift_prices[0] = 24.99
gift_prices
````

The price became 24, not 24.99!

> **Common mistake:** putting a float into an integer array. The dtype of `gift_prices` is `int64`, because we made
> it from whole numbers, and an array holds one type. So NumPy **cuts off** the decimal part of 24.99: it does not
> even round it to 25. There is no error and no warning. If an array may need decimals, make it from floats at the
> start: `np.array([25.0, 40.0, 15.0, 60.0])`.

## A slice is a view

Here is a surprise. Let's take a slice of `gift_prices`, and change an element of the slice:

````python
first_two = gift_prices[:2]
first_two[0] = 0
gift_prices
````

`gift_prices` changed too: its first price is now 0. A slice of an array is a **view**: a window onto the same
elements, not a copy of them. Changing the view changes the original array. NumPy works this way to save time and
memory with large arrays. (A slice of a *list* is different: it is a copy.)

When you want an independent array, add `.copy()`. First, let's put the price back to 25:

````python
gift_prices[0] = 25
first_two = gift_prices[:2].copy()
first_two[0] = 0
gift_prices
````

This time, `first_two` is a copy, and `gift_prices` keeps its price of 25.

## Arithmetic between two arrays

Now back to the six products of the shop. The revenue of a product is its units sold times its price. With two
arrays of the same length, one line computes the revenue of every product:

````python
revenue = units * prices
revenue
````

NumPy multiplies the two arrays **element by element**: the first element by the first element (10 × 12.00 = 120
for the coffee beans), the second by the second (25 × 4.50 = 112.5 for the tea), and so on. The result has one
revenue for each product. `units` holds integers and `prices` holds floats, so the result holds floats.

The other operators work the same way: `+`, `-`, `/` and `**`. For example, here is what each product costs the
shop, and the profit on one unit of each:

````python
costs = np.array([7.00, 2.00, 4.00, 18.00, 1.25, 6.50])
prices - costs
````

The shop earns $12.00 on each grinder it sells, but only $1.50 on a box of cookies.

To add up the revenue of all the products, we can use `.sum()` (the next chapter shows this and other summaries):

````python
revenue.sum()
````

The shop's revenue last week was $553.50.

> **Try it:** compute what the units sold last week cost the shop, for each product: `units * costs`.

## Arithmetic with a number

An array and a single number work together too: NumPy uses the number with every element. Here are the prices
during a sale with 10% off (we pay 90% of each price):

````python
prices * 0.9
````

The coffee beans now cost 10.80 dollars instead of 12.00, and every other price is 10% lower too. This is what we
did in the previous chapter with `prices * 1.08`, and in the chapter *What does a model learn?* with
`700 * customers["visits_per_month"]`. The other operators work with a number in the same way.

> **Try it:** raise every price by 50 cents with `prices + 0.50`. Then compute the units sold if every product
> sold 5 more, with `units + 5`.

## Arrays of different lengths

What if the two arrays do not have the same length? The manager wrote down this week's units for only the first
three products, and multiplies them by the six prices:

````python
this_week = np.array([12, 20, 9])
this_week * prices
````

The last line of the error message is `ValueError: operands could not be broadcast together with shapes (3,) (6,)`.
In plain words:

- the **operands** are the two arrays on the two sides of `*`;
- **broadcast** is NumPy's word for matching up the elements of the two operands;
- the **shapes** `(3,)` and `(6,)` say that one array has 3 elements and the other has 6.

NumPy cannot pair 3 elements with 6, and it does not guess which products the three numbers belong to. The fix is
to give it arrays of the same length: for example, the units of all six products, or a slice of the first three
prices, `this_week * prices[:3]`.

## NumPy functions for whole arrays

NumPy has many functions that work on every element of an array. Here are three.

**Rounding.** The shop also sells online in Europe. With an exchange rate of 0.9137 euros for one dollar, the
prices in euros have many decimals:

````python
euros = prices * 0.9137
euros
````

`np.round(a, 2)` rounds every element to 2 decimals, as prices should be:

````python
np.round(euros, 2)
````

The coffee beans cost 10.96 euros, the tea 4.11 euros, and so on.

**Absolute values.** Last week, the shop had a forecast of the units it would sell. The error of the forecast is the
real value minus the forecast (as in the chapter *What does a model learn?*):

````python
forecast = np.array([12, 20, 10, 2, 35, 8])
errors = units - forecast
errors
````

A negative error means that the forecast was too high: the shop expected 12 bags of coffee beans and sold 10, an
error of -2. Often we care only about how big an error is, not about its sign. `np.abs` gives the **absolute
value** of every element, the value without its sign:

````python
np.abs(errors)
````

The forecast was off by 2, 5, 2, 0, 5 and 2 units. The mean absolute error (MAE) of the chapter *A first
machine-learning example* is the average of such numbers.

**Square roots.** In the chapter *What does a model learn?*, we squared the errors. `np.sqrt` gives the square root
of every element, so it undoes a square:

````python
squared = errors ** 2
np.sqrt(squared)
````

The square roots are the sizes of the errors again: 2, 5, 2, 0, 5 and 2. The result holds floats (`2.` is 2.0),
because a square root is usually a decimal number: the square root of 2 is about 1.414.

## Summary

- **Indexing:** `a[0]` is the first element and `a[-1]` the last. Positions start at 0.
- **Slicing:** `a[start:stop:step]` gives several elements as an array: `a[1:4]`, `a[:3]`, `a[3:]`, `a[::2]`.
  **The stop is not included.**
- **Changing an element:** `a[0] = 25`. A float put into an integer array is cut off (24.99 becomes 24), without a
  warning.
- **A slice is a view:** changing it changes the original array. Use `a[:2].copy()` for an independent array.
- **Arithmetic between two arrays** of the same length works element by element: `units * prices` gives the
  revenue of each product. Arrays of different lengths give a `ValueError` (*could not be broadcast*).
- **Arithmetic with a number** works on every element: `prices * 0.9`, `prices + 0.50`.
- **Functions** on every element: `np.round(a, 2)`, `np.abs(a)`, `np.sqrt(a)`. `a.sum()` adds up all the
  elements.
