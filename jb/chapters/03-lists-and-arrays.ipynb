## Adding tax to a list of prices

A small shop sells five products. Here are their prices in a Python list. We want the price of each product with
8% sales tax added. With a list, we need a loop:

````python
prices_list = [20.00, 8.50, 12.25, 42.00, 4.50]

taxed_list = []
for price in prices_list:
    taxed_list.append(price * 1.08)
taxed_list
````

> **In Thonny:** a script does not show the value of the last line. Write `print(taxed_list)` instead. The same
> is true for every cell in this book whose last line shows a value (the chapter *Colab, Thonny and other IDEs*
> explains why).

Multiplying by 1.08 adds 8%: the first product costs 20.00, and 20.00 × 1.08 = 21.60. The loop takes the prices
one by one, computes each price with tax, and appends it to a new list.

If you know list comprehensions, you can write the same loop on one line. The result is the same:

````python
[price * 1.08 for price in prices_list]
````

Either way, we tell Python to go through the prices one at a time. Can we simply multiply the whole list by 1.08?

````python
prices_list * 1.08
````

No. Read the last line of the error message: Python *can't multiply sequence by non-int of type 'float'*. A list
is a *sequence*, and Python cannot multiply it by a decimal number (a `float`). A list does not know how to do
arithmetic on its values. (We see below what `*` does to a list with a whole number.)

## The same prices in a NumPy array

Now let's put the same prices into a NumPy **array**. `np.array` takes a list and makes an array from it:

````python
prices = np.array([20.00, 8.50, 12.25, 42.00, 4.50])
prices
````

Colab shows the array as `array([...])`. NumPy writes the numbers in a short form: `20.` means 20.0, and `8.5`
means 8.50. It also adds spaces, so that the numbers line up.

> **In Thonny:** `print(prices)` shows the values without the word `array` and without commas:
> `[20.    8.5  12.25 42.    4.5 ]`. It is the same array.

Now the tax:

````python
prices * 1.08
````

One short line, and no loop. `prices * 1.08` multiplies **every** value of the array by 1.08, and gives a new array
with the five results. They are the same numbers that the loop gave. The array `prices` itself does not change.

Arrays are also **fast**. NumPy keeps the numbers side by side in the computer's memory, all of the same type, and
it runs the loop for us in fast compiled code. With five prices, you cannot see the difference, but with millions
of values an array is many times faster than a Python loop. That is one reason why scikit-learn works with NumPy
arrays: in the chapter *A first machine-learning example*, the predictions that `predict` gave us were an array.

> **Try it:** change `1.08` to `1.10` to add 10% tax instead, and run the cell again.

## One object with many values

An array is **one** object that holds many values. Let's ask Python what kind of object `prices` is:

````python
type(prices)
````

Its type is `numpy.ndarray` (in Thonny, `print(type(prices))` shows `<class 'numpy.ndarray'>`). "nd" stands for
"n-dimensional": an array can also be a table, as the chapter *Two-dimensional arrays* shows. We simply say
"array". Each value in an array is an **element** of the array.

How many elements does `prices` have? `len` works as it does for a list. An array also knows its number of
elements by itself, as `.size`. With a comma between them, Colab shows both values, in parentheses:

````python
len(prices), prices.size
````

Both say 5. Notice that `.size` has no parentheses. It is not a function that we call, but a value that the array
carries with it, called an **attribute**. You saw another attribute in the first chapters: `customers.shape`. For
the arrays of this chapter, `len` and `.size` give the same number.

## The type of the values: dtype

An array also knows what type of values it holds, its **dtype** (short for "data type"):

````python
prices.dtype
````

Colab shows `dtype('float64')` (`print` shows just `float64`). **float64** means decimal numbers, or *floats*.
The 64 is the number of bits that each number uses in memory; you do not need to remember it.

Here are the units that the shop sold of each product today. They are whole numbers:

````python
units = np.array([3, 10, 1, 2, 25])
units.dtype
````

**int64** means whole numbers, or *integers*. What happens if we mix integers and floats in one array?

````python
mixed = np.array([3, 1.5, 4])
mixed
````

Every element became a float: `3.` and `4.` are 3.0 and 4.0. **An array holds one type of value.** When we mix
integers and floats, NumPy makes them all floats, because a float can hold every one of the values. A list is
different: in the list `[3, 1.5, 4]`, the 3 and the 4 stay integers.

An array can also hold text:

````python
regions = np.array(["North", "South", "East"])
regions
````

Colab shows the dtype at the end: `<U5` means text of up to 5 characters (the U is for Unicode, the standard way
to store text). We rarely need text arrays in this book, because pandas has its own way to handle text. Arrays
are mostly for numbers.

> **Common mistake:** typing a number in quotes, as in `np.array([12, "15", 9])`. One text value makes **all** the
> values text, and arithmetic on the array then fails. Write numbers without quotes.

## Lists and arrays behave differently

The same operator can do different things to lists and to arrays. Here are the units of three drinks that a café
sold in two weeks, as two lists:

````python
week1 = [30, 12, 25]
week2 = [28, 15, 20]
week1 + week2
````

For lists, `+` **joins** them: the result is one longer list with six values. Now the same with arrays:

````python
np.array(week1) + np.array(week2)
````

For arrays, `+` **adds** them element by element: the first value to the first value (30 + 28 = 58), the second to
the second (12 + 15 = 27), and so on. The result is the number of units of each drink in the two weeks together,
which is usually what a business wants.

`*` with a whole number also differs. First with the list:

````python
week1 * 2
````

`* 2` **repeats** the list: six values. Now with an array:

````python
np.array(week1) * 2
````

`* 2` **doubles** each value of the array. Here is a summary:

| Code | When `a` and `b` are lists | When `a` and `b` are arrays |
|---|---|---|
| `a + b` | joins `a` and `b` into one list | adds them element by element |
| `a * 2` | repeats `a` twice | doubles each element |
| `a * 1.08` | an error | multiplies each element by 1.08 |

> **Common mistake:** adding two lists and expecting sums. `[30, 12, 25] + [28, 15, 20]` gives a list of six
> values, with no error message to warn you. When you want arithmetic, make arrays first with `np.array`.

## Evenly spaced numbers: np.arange

`np.arange` makes an array of evenly spaced numbers, like Python's `range`. With one number, it starts at 0:

````python
np.arange(5)
````

Five numbers: 0, 1, 2, 3 and 4. **The stop value is not included**, as with `range`. To choose the start too,
write `np.arange(start, stop)`. Here are the month numbers of a year:

````python
months = np.arange(1, 13)
months
````

To get the months 1 to 12, the stop must be 13. A third number gives the **step**, the distance from one number to
the next. For example, a supplier sells boxes of 10, 20, 30, 40 or 50 units:

````python
np.arange(10, 60, 10)
````

It starts at 10, goes up in steps of 10, and stops before 60.

> **Common mistake:** `np.arange(1, 12)` for the months of a year. It stops at 11, so December is missing. The stop
> must be one step past the last number you want.

## From an array back to a list: tolist()

Sometimes we need a plain Python list again, for example for code that works only with lists. `.tolist()` turns an
array into a list:

````python
taxed = prices * 1.08
taxed.tolist()
````

The square brackets with commas, and no word `array`, show that this is a list. Its values are the same as those
of `taxed_list`, which the loop made at the start of the chapter.

> **Note:** `list(taxed)` also makes a list, but each value in it stays a NumPy number, and Colab shows each one in
> a long form, such as `np.float64(21.6)`. `.tolist()` gives plain Python numbers.

## Summary

- `import numpy as np` imports NumPy with the short name `np`.
- `np.array([20.00, 8.50, 12.25])` makes an **array** from a list. An array is one object (of type
  `numpy.ndarray`) that holds many values, its **elements**.
- Arithmetic with a number works on every element at once, with no loop: `prices * 1.08`. It is also fast.
- `len(a)` and `a.size` give the number of elements. `a.dtype` gives their type: `int64` (integers), `float64`
  (floats), or a text type such as `<U5`. An array holds **one type**: integers mixed with floats become floats.
- For lists, `+` joins and `* 2` repeats. For arrays, `+` adds element by element and `* 2` doubles each element.
- `np.arange(start, stop, step)` makes evenly spaced numbers. **The stop is not included.**
- `a.tolist()` turns an array back into a list.

### Question 8 (Challenge)

Write a function `add_tax(price_list, rate)`. It takes a list of prices and a tax rate (for example `0.08` for 8%),
uses a NumPy array to add the tax to every price, and returns a list. Test it on `prices_list` with the rate
`0.08`, and check that the result equals `taxed_list` from the start of the chapter.

<details class="yr-answer">
<summary>Answer</summary>

Inside the function, `np.array` makes an array from the list, one multiplication adds the tax, and `.tolist()`
turns the result back into a list. Adding a rate of 0.08 means multiplying by 1 + 0.08. The comparison with
`taxed_list` prints `True`.

```python
def add_tax(price_list, rate):
    with_tax = np.array(price_list) * (1 + rate)
    return with_tax.tolist()


result = add_tax(prices_list, 0.08)
print(result)
print(result == taxed_list)
```
