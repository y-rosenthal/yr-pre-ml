So far, every array was a single row of values: the prices of some products, or the visits of 500 customers. But
business data usually comes as a table, with rows and columns. In this chapter, we keep a table of numbers in a
two-dimensional NumPy array, select its rows and columns, and summarize it row by row or column by column. At the
end, we see why this matters for machine learning: the predictors of a model are exactly such a table.

**In this chapter, you will learn to:**

- create a two-dimensional array from a list of lists, and read its `.shape` (rows, columns) and `.ndim`;
- select one value with `a[row, col]`, a whole row or column, and a block of rows and columns with slices;
- summarize each column with `axis=0`, and each row with `axis=1`;
- do arithmetic on a whole table with one number;
- explain why, in machine learning, `X` is a 2-D array and `y` a 1-D array, and why `X.shape[0]` must equal
  `y.shape[0]`.

````python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

DATA_URL = "https://raw.githubusercontent.com/y-rosenthal/yr-pre-ml/main/data/"
````

We need pandas and scikit-learn only in the last sections, about machine learning.

## A table of numbers

A chain has four stores, S1 to S4. Here are the units of its best-selling product that each store sold in the
first six months of the year:

| Store | Jan | Feb | Mar | Apr | May | Jun |
|---|---|---|---|---|---|---|
| S1 | 150 | 160 | 140 | 175 | 155 | 120 |
| S2 | 80 | 90 | 85 | 95 | 90 | 100 |
| S3 | 140 | 150 | 175 | 170 | 190 | 195 |
| S4 | 60 | 65 | 70 | 60 | 75 | 90 |

To put this table into NumPy, we write it as a list of rows. Each row is a list of six numbers, and the four rows
go inside one more pair of square brackets:

````python
units = np.array([
    [150, 160, 140, 175, 155, 120],
    [80, 90, 85, 95, 90, 100],
    [140, 150, 175, 170, 190, 195],
    [60, 65, 70, 60, 75, 90],
])
units
````

> **In Thonny:** wrap the last line in `print()`: `print(units)`. `print()` shows the same table without
> `array(...)` and without commas, one row on each line.

NumPy shows one store on each line, and lines up the columns. This is a **two-dimensional array** (a **2-D
array**): its values go in two directions, down (the stores) and across (the months). The arrays of the earlier
chapters were **one-dimensional** (1-D): a single row of values.

The names of the stores and of the months are not in the array: it holds only the numbers. We must remember that
row 0 is store S1 and column 0 is January. (pandas, from the next chapter on, keeps such names with the values.)

## The shape: rows and columns

````python
print("shape:", units.shape)
print("ndim: ", units.ndim)
print("size: ", units.size)
````

The **shape** of an array gives its size in each direction. For a 2-D array, it is always **(rows, columns)**: here
4 rows (the stores) and 6 columns (the months). `.ndim` is the **number of dimensions**: 2. And `.size` is the
number of values: 4 × 6 = 24.

For a 1-D array, the shape has only one number: `np.array([5, 7, 9]).shape` is `(3,)`, and its `.ndim` is 1. The
comma in `(3,)` tells us that this is a tuple with one value, not the number 3 in parentheses. You saw such shapes,
`(3,)` and `(6,)`, in an error message in the chapter *Indexing, slicing and arithmetic*.

## One value: `a[row, col]`

To get one value, give its row and its column, in that order, separated by a comma. How many units did store S2
(row 1) sell in March (column 2)?

````python
units[1, 2]
````

85 units. (Colab shows `np.int64(85)`: the number 85, as in the chapter *Indexing, slicing and arithmetic*.) As
in 1-D arrays, the positions start at 0: row 1 is the second store, and column 2 is the third month. Negative
positions count from the end, so `units[-1, -1]` is the last store in the last month: 90.

> **Common mistake:** swapping the row and the column. `units[2, 1]` is store S3 in February (150), not store S2 in
> March. The row always comes first.

## A whole row or a whole column

With only one position, we get a whole row. Here are the six months of store S2:

````python
units[1]
````

For a column, we need a way to say "all the rows". That is a colon alone, `:`: a slice with no start and no stop,
which takes everything. Here is March, for all four stores:

````python
units[:, 2]
````

Read `units[:, 2]` as "all rows, column 2". In March, S3 sold the most (175) and S4 the least (70). In the same
way, `units[1, :]` means "row 1, all columns", the same as `units[1]`. A single row or a single column is a 1-D
array.

## Slicing rows and columns

A slice can choose rows and columns at once. The rules of 1-D slices apply in each direction: the start is
included, and the stop is not. Here are the first two stores (rows 0 and 1) in the first three months (columns 0,
1 and 2):

````python
units[0:2, 0:3]
````

The result is a smaller table: a 2-D array with the shape (2, 3). A slice with no start begins at the first row or
column, and a slice with no stop goes to the end. So `units[:, 3:]` is all four stores from April to June, the
second quarter of the year.

## Summaries along an axis

`sum()` with no argument adds up every value of the table:

````python
units.sum()
````

The four stores sold 2,880 units in six months. But a manager usually wants a total for each month, or for each
store. For that, we tell `sum` in which direction to add up, with the argument `axis`.

**`axis=0` adds down the rows**: it gives one total for each column, here for each month.

````python
units.sum(axis=0)
````

In January, the four stores sold 150 + 80 + 140 + 60 = 430 units together. The chain's sales grew until May (510),
then dipped a little in June (505).

**`axis=1` adds across the columns**: it gives one total for each row, here for each store.

````python
units.sum(axis=1)
````

Store S1 sold 900 units in six months, and store S3, the best store, sold 1,020. This table shows both results
around the data:

| | Jan | Feb | Mar | Apr | May | Jun | `sum(axis=1)` → |
|---|---|---|---|---|---|---|---|
| **S1** | 150 | 160 | 140 | 175 | 155 | 120 | **900** |
| **S2** | 80 | 90 | 85 | 95 | 90 | 100 | **540** |
| **S3** | 140 | 150 | 175 | 170 | 190 | 195 | **1020** |
| **S4** | 60 | 65 | 70 | 60 | 75 | 90 | **420** |
| **`sum(axis=0)`** ↓ | **430** | **465** | **470** | **500** | **510** | **505** | **2880** |

A way to remember it: the shape is (4, 6), and **the axis you name is the one that disappears**. Axis 0 is the
first number of the shape, the 4 rows: `axis=0` adds them up and leaves 6 results, one for each column. Axis 1 is
the 6 columns: `axis=1` adds them up and leaves 4 results, one for each row.

`mean`, `min` and `max` take `axis` too. Here are the average units of each store in a month, and the best result
of any store in each month:

````python
print(units.mean(axis=1))
print(units.max(axis=0))
````

In an average month, store S3 sells 170 units, and store S4 only 70. The best result of a single store rose from
150 units in January to 195 in June. (`max` gives the largest value, not which store it belongs to; a practice
question finds the store with `argmax`.)

> **Common mistake:** mixing up the axes. If you want the total of each of the 4 stores and get 6 numbers, you
> used `axis=0` instead of `axis=1`. Check the length of the result.

## Arithmetic with a number

Arithmetic with one number works on every value of a 2-D array, as it does on a 1-D array. The product sells for
$25. Here is the revenue of each store in each month:

````python
revenue = units * 25
revenue
````

The result has the same shape as `units`, (4, 6): store S1 made 150 × 25 = $3,750 in January, and so on. All the
tools of this chapter work on the new table too: for example, `revenue.sum(axis=1)` gives the revenue of each
store.

## The bridge to machine learning: `X` and `y`

Why do 2-D arrays matter? Because the data of a machine-learning model is a table. Think of the customers of the
chapter *A first machine-learning example*:

- The **predictors** form a table, with **one row for each example** (a customer) and **one column for each
  predictor**. That is a 2-D array, named `X`.
- The **target** has **one value for each example** (what the customer spent). That is a 1-D array, named `y`.

(The capital `X` is a custom from mathematics, where a capital letter often names a table of numbers, and a small
letter a single list of numbers.)

Let's make `X` and `y` from `customers.csv`, with two predictors: how many months each person has been a customer,
and how often they visit a store.

````python
customers = pd.read_csv(DATA_URL + "customers.csv")
X = customers[["months_as_customer", "visits_per_month"]].to_numpy()
y = customers["annual_spend"].to_numpy()
X.shape, y.shape
````

These lines use pandas, which the next chapter, *Series and DataFrames*, starts to teach. In short: the first line
reads the data file into a table, as in that chapter. The second line takes two columns of the table (the
double brackets hold a list of column names), and `.to_numpy()` turns that table into a NumPy array. The last
line takes one column, and `.to_numpy()` turns it into a 1-D array. So **`df.to_numpy()` turns a pandas table into
a NumPy array.**

The shapes say it all. `X` has the shape (500, 2): 500 rows (the customers) and 2 columns (the predictors). `y` has
the shape (500,): 500 values, one for each customer. Let's look at the first three customers:

````python
print(X[:3])
print(y[:3])
````

With only one slice, `X[:3]` takes the first three rows, with all their columns. (NumPy shows `8.` for 8 months: an
array holds one type, and with the floats of `visits_per_month`, the months became floats too.)

Row 0 of `X` and `y[0]` describe the same customer: a member for 8 months, who visits 5.6 times a month, and who
spent $3,577.54 in a year. Row 1 and `y[1]` are the second customer, and so on. A model learns from these pairs.
So **the number of rows of `X` must equal the number of values of `y`**: `X.shape[0] == y.shape[0]`, where
`shape[0]` is the first number of the shape.

What if they do not match? Let's give a model all 500 rows of `X`, but only the first 400 values of `y`:

````python
model = LinearRegression()
model.fit(X, y[:400])
````

The last line is `ValueError: Found input variables with inconsistent numbers of samples: [500, 400]`. A *sample*
is scikit-learn's word for an example, one row. Before it learns anything, `fit` checks the shapes: `X` has 500
rows, but `y` has only 400 values, so 100 customers have no target. The model refuses.

With all of `y`, the shapes match, and `fit` works on our NumPy arrays:

````python
model = LinearRegression()
model.fit(X, y)
np.round(model.coef_, 2)
````

As in the chapter *What does a model learn?*, linear regression learned one number for each predictor, so one
number for each **column** of `X`: about $6 of yearly spending for each month as a customer, and about $792 for
each visit per month.

one column.

The next chapters teach pandas, the tool that builds `X` and `y` from a real table: choosing columns, filtering
rows, filling in missing values, turning text into numbers. When the table is ready, `.to_numpy()` gives the arrays
of this chapter.

## Summary

- A **2-D array** holds a table of numbers: `np.array([[...], [...]])`, a list of rows, all of the same length.
- `.shape` gives **(rows, columns)**, and `.ndim` the number of dimensions (2). A 1-D array has a shape such as
  `(500,)`.
- `a[row, col]` is one value. `a[1]` (or `a[1, :]`) is a whole row, and `a[:, 2]` a whole column. Slices work in
  both directions: `a[0:2, 0:3]`.
- `a.sum(axis=0)` adds down the rows: one result for each **column**. `a.sum(axis=1)` adds across the columns: one
  result for each **row**. `mean`, `min`, `max` (and `argmax`) take `axis` too.
- Arithmetic with a number, such as `units * 25`, works on every value, and keeps the shape.
- In machine learning, `X` is a 2-D array (one row for each example, one column for each predictor), and `y` is a
  1-D array (one value for each example). `X.shape[0]` must equal `y.shape[0]`. `df.to_numpy()` turns a pandas
  table into a NumPy array.
