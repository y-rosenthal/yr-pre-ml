## CSV files

The most common file format for tables is **CSV**, short for *comma-separated values*. A CSV file is plain text:
one line for each row, with the values separated by commas. The first line holds the column names. Here are the
first lines of the file `sales.csv`:

```
date,store,region,category,units,price,discount
2025-01-01,S1,North,Clothing,3,46.99,0.1
2025-01-01,S8,West,Grocery,9,2.99,0.0
2025-01-01,S2,North,Grocery,11,5.99,0.0
```

Each line is one sale of a store chain: the day, the store and its region, the product category, the number of
units, the price of one unit in dollars, and the discount as a fraction of the price (`0.1` is 10%). The appendix
*The datasets* describes every file and column.

## Reading a CSV file

`pd.read_csv` reads a CSV file and returns a DataFrame:

````python
sales = pd.read_csv(DATA_URL + "sales.csv")
sales.head()
````

`DATA_URL + "sales.csv"` joins two strings into the full address of the file. `pd.read_csv` reads the file, takes
the column names from the first line, and makes one row of the DataFrame from each other line. It also chooses a
dtype for each column, from the values in it.

`head()` shows the first five rows. It is the first thing to look at after reading a file: did the columns come out
right, and what do the values look like?

> **Note:** to read a file that is saved on your own computer, give its name instead of the address:
> `pd.read_csv("sales.csv")` reads the file `sales.csv` in the folder of your Thonny script. The appendix
> *The datasets* has a link to each file.

## A report on the table: `info()`

`info()` gives the facts of the last sections, and a little more, in one report:

````python
sales.info()
````

Let's read it from the top:

- The first line says that `sales` is a DataFrame.
- `RangeIndex: 2000 entries, 0 to 1999`: there are 2,000 rows, with the index 0 to 1999.
- `Data columns (total 7 columns)`, then one line for each column: its position (`#`), its name, its **non-null
  count** and its dtype.
- The `dtypes` line counts the columns of each dtype.
- The last line says how much memory the table uses (about 110 KB).

The **non-null count** is the number of values that are not missing ("null" is another word for a missing value).
Here, every column has 2000 non-null values, as many as there are rows: no value is missing in `sales.csv`. We meet
missing values at the end of this chapter.

> **Note:** `info()` prints its report itself, so in Thonny, too, you write `sales.info()` without `print()`. With
> `print(sales.info())`, you see the report and then the word `None`: `info()` prints, but it returns no value.

## Summary statistics: `describe()`

`describe()` computes summary statistics for every number column:

````python
sales.describe()
````

The text columns are not there: `describe()` uses only the number columns. Each row of the result is one
statistic:

- **count**: the number of values that are not missing; 2000 in each column.
- **mean**: the average. A sale has 4.47 units on average, and the average price is $72.83.
- **std**: the *standard deviation*, a typical distance of the values from the mean. A small std means that the
  values are close together, and a large one that they are spread out.
- **min** and **max**: the smallest and the largest value. A sale has from 1 to 16 units, and the prices go from
  $1.99 to $597.99.
- **25%, 50%, 75%**: the *quartiles*. Sort the values from the smallest to the largest: the 25% value is a quarter
  of the way through the list, the 50% value halfway, and the 75% value three quarters of the way. The 50% value is
  the **median**, the middle value: half of the values are at or below it.

What do these numbers tell a manager?

- **Prices:** the median price is $26.99, but the mean is $72.83. So half of the sales have a price of $26.99 or
  less, and a few expensive items (up to $597.99) pull the mean up. When the mean and the median are far apart, the
  median is often the better "typical" value. The large std of the prices (about 122) says the same thing: the
  prices are very spread out.
- **Units:** the 75% value is 6, so at least three quarters of the sales have 6 units or fewer. Sales of 16 units
  are rare.
- **Discounts:** the 25% and the 50% values are 0, so at least half of the sales had no discount.

> **Try it:** the numbers have many decimals. Run `sales.describe().round(2)` to round them to two decimals.

## Text columns: `unique()` and `nunique()`

For a text column, a mean or a median makes no sense. Instead, we ask: which different values does the column
have? `unique()` lists each value once:

````python
sales["store"].unique()
````

The eight stores, S1 to S8, in the order in which they first appear in the file. In pandas 2 (Colab), the result
is a NumPy array with `dtype=object`; pandas 3 shows a `<StringArray>` with the same values. `nunique()` counts the
different values:

````python
sales["store"].nunique()
````

There are 8 stores. When a column has many different values, this count is easier to read than the list. (The
chapter *Grouping and summarizing* counts how many sales each store had.)

## Missing values: a first look

The file `customers.csv`, from the chapter *A first machine-learning example*, describes 500 customers. Let's read
it and look at its `info()`:

````python
customers = pd.read_csv(DATA_URL + "customers.csv")
customers.info()
````

There are 500 rows (`RangeIndex: 500 entries`) and 6 columns. But two columns have **fewer than 500** non-null
values: `age` has 486, and `avg_basket` (the average amount spent per visit) has 489. So 14 ages and 11 basket
amounts are missing. pandas shows a missing value as `NaN` ("Not a Number"). `info()` is the quickest way to see
which columns have missing values, and how many.

The count row of `describe()` shows the same thing:

````python
customers[["age", "avg_basket"]].describe()
````

The count of `age` is 486, not 500: `describe()` skips the missing values, and computes the other statistics from
the values that are there. The average customer is about 46 years old and spends about $51.54 per visit.

Missing values are a problem for machine learning: many models, such as `LinearRegression`, refuse them. That
is why the chapter *A first machine-learning example* dropped those rows before `fit`. The chapter *Preparing data
for machine learning* shows how to drop them or fill them in.

## Summary

- `pd.read_csv(DATA_URL + "sales.csv")` reads a CSV file into a DataFrame.
- `head()` and `tail()` show the first and the last five rows; `head(10)` shows the first ten.
- `.shape` gives (rows, columns), `.columns` the column names, and `.dtypes` the type of each column (text is
  `object`, or `str` in pandas 3).
- `info()` prints the number of rows, and for each column its non-null count and its dtype. A non-null count below
  the number of rows means that values are missing.
- `describe()` gives the count, mean, std, min, quartiles (25%, 50% = the median, 75%) and max of each number
  column.
- For a text column, `unique()` lists the different values, and `nunique()` counts them.
