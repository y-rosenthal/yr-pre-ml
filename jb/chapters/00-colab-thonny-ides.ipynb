## Two ways to run the code

**Google Colab** (<https://colab.research.google.com>) is a free service from Google. It runs Python on Google's
computers, and you use it in your web browser. You need a Google account, but you install nothing. Colab already
has the packages this book uses: NumPy, pandas, scikit-learn and matplotlib.

**Thonny** (<https://thonny.org>) is the Python editor that you may know from *Think Python*. It runs Python on
your own computer. You install it once, and then you install the packages once (we show how below).

The book is written for Colab first. All the code also runs in Thonny, with one small change that this chapter
explains.

## Running a chapter in Colab

To open a chapter in Colab, click the **Open in Colab** button at the top of the chapter. On the website, the
rocket button at the top of the page does the same: choose **Colab**.

Colab shows the chapter as a list of *cells*. A text cell holds explanations. A code cell holds Python code, with
a small "play" button on its left. To run a code cell, click in it and press **Shift+Enter** (or click the play
button). The result appears under the cell.

Try it with the cell below. It computes the price of three items that cost $19.99 each.

````python
price = 19.99
quantity = 3
price * quantity
````

Colab shows `59.97` under the cell. Notice that we did not write `print()`. **When the last line of a cell is an
expression, Colab shows its value.** We use this a lot in this book, because it keeps the code short.

> **Note:** run the cells in order, from the top. A cell can use the variables that earlier cells made. If you
> skip a cell, a later cell may fail with a `NameError`. In Colab, **Runtime > Run all** runs every cell, in order.

**Save your own copy.** The chapter that Colab opens belongs to the book. If you change it and close the tab, your
changes are lost. To keep them, choose **File > Save a copy in Drive**. Colab saves the copy in your Google Drive,
and you can open it again later.

## The one difference: showing a value

Here is the same calculation again, with the result in a variable:

````python
total = price * quantity
total
````

In Colab, the last line, `total`, shows `59.97`.

Thonny runs a whole file of code as a *script*. In a script, a line such as `total` computes the value, and then
nothing happens: **a script shows only what you print**. So in Thonny, write the last line inside `print()`:

```python
total = price * quantity
print(total)
```

That is the only change to the code that you need to make. When a cell in this book ends with an expression whose
value Colab shows, wrap that expression in `print()` in Thonny.

> **In Thonny:** you can also type an expression in Thonny's **Shell** (the lower part of the window) after
> your script has run. The Shell shows the value, just as Colab does. For example, run the script, then type
> `total` in the Shell and press Enter.

You can use `print()` in Colab too. It shows the value in plain text:

````python
print(total)
````

## Getting Thonny ready

Thonny comes with its own copy of Python. The packages of this book are not part of it, so install them once:

1. In Thonny, choose **Tools > Manage packages...**.
2. Type `numpy` in the search box and click **Search on PyPI**. Click `numpy` in the results, then click
   **Install**.
3. Do the same for `pandas`, `scikit-learn` and `matplotlib`.

> **Common mistake:** the package is *installed* as `scikit-learn`, but *imported* as `sklearn`:
> `from sklearn.linear_model import LinearRegression`. Searching for `sklearn` in Thonny does not find the right
> package.

To use the code of a chapter in Thonny, copy the code cells into a new file in Thonny, one after the other, and
add `print()` where you need it. Leave out the cells that show an error on purpose (the text right after each one
explains the error). In a script, an error stops the program, and the lines after it do not run. Save the file,
then click **Run** (the green arrow, or F5).

## Which versions do you have?

Packages change over time. The code in this book was checked with the versions that Colab has, and also with the
newest versions. This cell shows the versions that *your* Python uses:

````python
import sys
import numpy
import pandas
import sklearn
import matplotlib

print("Python:      ", sys.version.split()[0])
print("numpy:       ", numpy.__version__)
print("pandas:      ", pandas.__version__)
print("scikit-learn:", sklearn.__version__)
print("matplotlib:  ", matplotlib.__version__)
````

On this website, the output shows the versions that the book was built with, which are Colab's versions. Run the
cell in Thonny and compare.

- If your versions are the **same or newer**, you are ready.
- If one of the packages is **older**, upgrade it: in **Tools > Manage packages...**, search for the package,
  click it, and click **Upgrade**. (If the Upgrade button is greyed out, you already have the newest version. If
  you click Upgrade and the version does not change, your Python is too old for the newest version: see the note
  below.)

> **Note:** in Thonny, choose **Help > About Thonny** to see your versions of Thonny and of its Python. The newest
> versions of NumPy and pandas need a recent Python. If Thonny cannot upgrade a package, install the newest Thonny
> from <https://thonny.org>. The book was checked with Colab's versions and with the newest ones (pandas 3). A
> somewhat older version usually works too, but some outputs may look a little different.

## Charts in Thonny

Some chapters draw charts with matplotlib. Every chart cell ends with `plt.show()`. In Colab, the chart appears
under the cell. In Thonny, `plt.show()` opens the chart in a new window, and **your program waits until you close
that window**. Close it to let the rest of the script run.

## Other editors

You can use other tools too. The same rule applies to all of them: a notebook shows the value of the last line, a
script shows only what you print.

- **Jupyter** (JupyterLab or Jupyter Notebook) on your own computer works like Colab. Install the packages first
  (for example `pip install numpy pandas scikit-learn matplotlib` in a terminal).
- **VS Code** opens `.ipynb` notebooks with its Jupyter extension: then it works like Colab. In a `.py` file, it
  works like Thonny.
- **Spyder** runs scripts like Thonny. Its console (the IPython console) shows the value of an expression that you
  type there, like Thonny's Shell.

To open a chapter as a notebook in one of these tools, download it: in Colab, choose **File > Download > Download
.ipynb**.

## When something fails

When Python cannot run a line, it shows an *error message*. A long message can look frightening, but **the last
line is the important one**: it names the kind of error and gives a short reason. Here are the errors you are most
likely to meet while you set up:

- `ModuleNotFoundError: No module named 'pandas'`: the package is not installed in the Python that runs your code.
  In Thonny, install it with **Tools > Manage packages...**.
- `NameError: name 'customers' is not defined`: a cell or line that makes `customers` did not run. Run the cells in
  order, from the top.
- An error that mentions `URL`, `HTTP`, `SSL` or `connection`: the data could not be read from the internet. Check
  that you are online, or use a copy of the file (see above).

The appendix *Common errors and how to fix them* has more.
